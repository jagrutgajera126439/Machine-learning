```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Select only actual numeric columns for rescaling
numeric_cols = ['tweet_id', 'airline_sentiment_confidence', 'negativereason_confidence', 'retweet_count']

# Fill missing values in numeric columns (e.g., negativereason_confidence) with their median
x = df[numeric_cols].copy()
for col in x.columns:
    x[col] = x[col].fillna(x[col].median())

# Rescale the numeric data
scaler = MinMaxScaler()
rescaled_data = scaler.fit_transform(x)

# Display Results
print(rescaled_data)
```

```python
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder object
encoder = LabelEncoder()

# Encode an existing categorical column from the Twitter dataset, such as 'airline'
df['airline_encoded'] = encoder.fit_transform(df['airline'])

# Display encoded data
print(df[['airline', 'airline_encoded']].head())
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Select numeric features
X = df[['tweet_id', 'airline_sentiment_confidence', 'negativereason_confidence', 'retweet_count']]

# Handle missing values
X = X.fillna(X.mean())

# Standardize data
X_scaled = StandardScaler ().fit_transform(X)

# Apply PCA
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled)

# Plot
plt.scatter(principal_components[:, 0], principal_components [ :, 1])
plt.xlabel("Principal Component 1")
plt.ylabel ("Principal Component 2")
plt.title("PCA")
plt.show()
```
